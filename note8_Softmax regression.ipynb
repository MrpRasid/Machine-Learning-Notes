## Softmax regression 

**Softmax regression (or multinomial logistic regression):** <br>
is a machine learning algorithm used for multi-class classification. It generalizes standard binary logistic regression to handle scenarios with three or more mutually exclusive classes. The model calculates a score for each class and maps them into a valid probability distribution that sums to one.
