## Lasso Regression (Least Absolute Shrinkage and Selection Operator):
Lasso Regression is one of the most important regularization techniques in Machine Learning. It is used to solve one of the biggest problems in Linear Regression—overfitting.

If Ridge Regression reduces coefficients, Lasso Regression can reduce them all the way to zero, automatically selecting the most important features.

- A normal Linear Regression model may still assign small coefficients to these useless features.

Lasso automatically removes them.

- Lasso can be use for feature selection automatically
