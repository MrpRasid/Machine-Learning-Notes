## Elastic Net Regression

Elastic Net Regression is a machine learning technique used to predict continuous values. It combines two other regression methods, `Ridge` and `Lasso`, to prevent model overfitting. It handles datasets with many features and highly correlated variables very effectively.

ElasticNet Regression based on two paramete:
- **Alpha (α):** Controls the overall strength of the penalty applied to the model.
- **L1 Ratio:** Determines the exact mix between the Lasso and Ridge penalties. An L1 ratio of 1 means strictly Lasso, while a ratio of 0 means strictly Ridge

### Primary Benefits
**Feature Selection:** Automatically removes unnecessary data features. <br>
**Multicollinearity:** Keeps or removes related groups of data features together rather than making random choices. <br>
**Stability:** Prevents the model from memorizing training data too closely.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x,y = load_diabetes(return_X_y = True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
# Linear Regression 
reg = LinearRegression()
reg.fit(x_train, y_train)
y_pred = reg.predict(x_test)
r2_score(y_test, y_pred)
```

```python
# Ridge Regression 
ridg = Ridge(alpha = 0.1)
ridg.fit(x_train, y_train)
y_predict1 = ridg.predict(x_test)
r2_score(y_test, y_predict1)
```

```python
## Lasso Regression 
lso = Lasso( alpha = 0.01)
lso.fit(x_train, y_train)
y_predict2 = lso.predict(x_test)
r2_score(y_test, y_predict2)
```

```python
## ElasticNet
Els = ElasticNet(alpha = 0.005, l1_ratio = 0.9)
Els.fit(x_train, y_train)
y_predict3 = Els.predict(x_test)
r2_score(y_test, y_predict3)
```
