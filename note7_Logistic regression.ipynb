## Logistic regression

Logistic regression is a supervised machine learning algorithm used for classification tasks. Instead of predicting continuous numbers, it calculates the probability that an observation belongs to a specific category (e.g., yes/no or 0/1).

Line equation:

$$ Ax_1 + Bx_2 + C = 0 $$
or $$ W_0 + W_1 x_1 + W_2 x_2 = 0 $$
or $$ W_0 x_0 + W_1 x_1 + W_2 x_2 = 0 $$
 or $$ \sum_{i=1}^{n} W_i x_i = 0$$

## use perceptron trick 

```python
from sklearn.datasets import make_classification 
import numpy as np
```

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.figure(figsize=(7,4))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1
    for i in range(1000):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        weights = weights + lr*(y[j]-y_hat)*X[j]
        
    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(7,4))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

## Sigmoid Function 

The sigmoid function, also known as the logistic function, is a mathematical function widely used in machine learning to map any real-valued number to a value between 0 and 1.

Formula:

$$ \sigma({z})=\frac{1}{1+e^{-z}} $$ 

how it's work with sigmoid

## perceptron using sigmoid function:

```python
def perceptron(x,y):
    x = np.insert(x,0,1, axis = 1)
    weights = np.ones(x.shape[1])
    lr = 0.1
    for i in range(2000):
        j = np.random.randint(0,100)
        y_hat = sigmoid(np.dot(x[j], weights))
        weights = weights - (lr * (y[j] -y_hat) * x[j])

    return weights[0], weights[1:]
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
intercept_,coef_ = perceptron(X,y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input2 = np.linspace(-3,3,100)
y_input2 = m*x_input + b
```

```python
plt.figure(figsize=(7,4))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.plot(x_input1,y_input1,color='black',linewidth=3)
plt.plot(x_input2,y_input2,color='brown',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

* brown line define the perceptron using sigmoid function 

## Use Gradient Descent 

weight's using GD 



```python
def gd(X,y):
    
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.5
    
    for i in range(5000):
        y_hat = sigmoid(np.dot(X,weights))
        weights = weights + lr*(np.dot((y-y_hat),X)/X.shape[0])
        
    return weights[1:],weights[0]
        
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
coef_,intercept_ = gd(X,y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input1 + b
```

```python
plt.figure(figsize=(7,4))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.plot(x_input1,y_input1,color='black',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
