# Gradient Descent:
Gradient descent is an iterative optimization algorithm used to minimize a model's error by continuously adjusting its internal parameters

### Core Mechanics & Step-by-Step Process

- **Initialization**: The model starts with random values for its parameters (weights and biases).
- **Calculate Loss**: The Loss Function evaluates how far off the model's predictions are from the actual values.
- **Compute the Gradient**: Calculus is used to find the partial derivatives (the gradient) of the loss function. This acts as a compass, pointing in the direction of the steepest increase in error
- Update Parameters: The algorithm takes a step in the opposite direction of the gradient to reduce the error.<br>
    Update formuls for Loss/Error:
                   $$\text{Error} = y_i - (mx_i + b)$$
                  $$b_{\text{new}} = b_{\text{old}} - \eta \cdot \frac{\partial L}{\partial b}$$
  
                  $$m_{\text{new}} = m_{\text{old}} - \eta \cdot \frac{\partial L}{\partial m}$$
  
- Repeat: Steps 2 through 4 are repeated for hundreds or thousands of iterations until the slope becomes flat, meaning the lowest point has been found.

**What are the algorithm use `Gradient Descent`**?
- Linear Regression
- Logistic Regression
- t-SNE

Also Gradient Descent Most important in Deep Learning

### Types of Gradient Descent
- Batch Gradient Descent
- Stochastic Gradient Descent (SGD)
- Mini-Batch Gradient Descent (MBGD)

### Batch Gradient Descent: <br>
`Batch Gradient Descent is an optimization algorithm that calculates the error for the entire training dataset before updating the model's parameters.` It is the standard form of gradient descent, where the term "batch" refers to processing the complete set of available training data all at once during each step

### Stochastic Gradient Descent (SGD):
is a machine learning optimization algorithm that updates model parameters efficiently by processing one training example at a time

- It makes learning much faster and less memory-intensive than traditional batch gradient descent, while its inherent randomness helps models escape suboptimal solutions/ local minima.
- Looks at just one single random row of data, makes an update, and repeats. (Incredibly fast, but the path to the best solution is chaotic and jumps all over the place).

**When to use Stochastic Gradient Descent(SGD)?**

- `Massive Datasets:` Batch gradient descent is too slow because it processes all data at once. SGD processes one random sample at a time.
- selecting datas are random

## Mini Batch Gradient descent

Splashes right down the middle. It splits the dataset into small groups called batches (usually powers of 2, like 32, 64, or 128 rows). It calculates the error and updates the model after looking at each batch.
